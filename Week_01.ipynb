```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt

# Membuat list untuk menyimpan url dari beberapa image
urls =["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
"https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
"http://farm3.staticflickr.com/2701/4247121446_0003bc2d90_z.jpg"]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
 image = io.imread(url) #read image
 image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
 image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) #convert color to RGB
 final_frame = cv.hconcat((image, image_2)) #concatenate image
 cv2_imshow(final_frame) #show image
 print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = "," x ",lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
 image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import random

img = cv2.imread("/content/drive/MyDrive/PCVK_2026/Student_KTM.jpeg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# --- SOAL 1: Penggunaan figsize ---
plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.title(f"1. Gambar Asli (Resolusi: {img_rgb.shape[1]}x{img_rgb.shape[0]})")
plt.show()
```

```python
#

#SOAL 2: Channel Red-Blue dan Green-Blue
# Red-Blue (Green dinolkan)
img_rb = img_rgb.copy()
img_rb[:,:, 1] = 0

# Green-Blue (Red dinolkan)
img_gb = img_rgb.copy()
img_gb[:,:,0] = 0

fig, ax = plt.subplots(1, 2, figsize=(12, 6))
ax[0].imshow(img_rb)
ax[0].set_title('2a. Red-Blue Channel')
ax[1].imshow(img_gb)
ax[1].set_title('2b. Green-Blue Channel')
plt.show()
```

```python
crop_img = img_rgb[20:115, 25:120]
plt.figure(figsize=(4, 4))
plt.imshow(crop_img)
plt.title('3. Crop (Baris 20-115, Kolom 25-120)')
plt.show
```

```python
# --- SOAL 4: Baris 5-30, semua kolom, Channel Red saja ---
# Agar citra tetap terlihat merah (bukan grayscale), kita buat matriks kosong dan hanya mengisi channel Red
img_red_only = np.zeros_like(img_rgb[5:30, :])
img_red_only[:,:,0] = img_rgb[5:30, :, 0]

plt.figure(figsize=(10, 3))
plt.imshow(img_red_only)
plt.title('4. Crop Baris 5-30, Channel Red Saja')
plt. show()
```

```python
img_flipped = cv2.flip(img_rgb, 0)
plt.figure(figsize=(10, 6))
plt.imshow(img_flipped)
plt.title('6. Citra Posisi Terbalik')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import random

img = cv2.imread("/content/drive/MyDrive/PCVK_2026/Student_KTM.jpeg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

img_boxes = img_rgb.copy()
h, w, _ = img_boxes.shape

for _ in range(5):
    # Buat koordinat dan warna secara acak
    x1 = random.randint(0, w - 100)
    y1 = random.randint(0, h - 100)
    x2 = min(x1 + random.randint(30, 150), w - 1)
    y2 = min(y1 + random.randint(30, 150), h - 1)
    color = (random.randint(0, 255), random.randint(0, 255), random.randint(0, 255))
    thickness = random.randint(3, 8)

    cv2.rectangle(img_boxes, (x1, y1), (x2, y2), color, thickness)

plt.figure(figsize=(10, 6))
plt.imshow(img_boxes)
plt.title("5. Lima Kotak Random")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img_aktivitas = cv2.imread("/content/drive/MyDrive/PCVK_2026/Student_KTM.jpeg")
img_act_rgb = cv2.cvtColor(img_aktivitas, cv2.COLOR_BGR2RGB)

# Position and size of the face
x_face = 300
y_face = 800
w_face = 130
h_face = 170

# Draw rectangle around the face
cv2.rectangle(
    img_act_rgb,
    (x_face, y_face),
    (x_face + w_face, y_face + h_face),
    (0, 255, 0),
    4
)

# Calculate center of the face
center_x_face = x_face + (w_face // 2)
center_y_face = y_face + (h_face // 2)

# Draw circle on the face
radius_face = 65

cv2.circle(
    img_act_rgb,
    (center_x_face, center_y_face),
    radius_face,
    (255, 0, 0),
    4
)

plt.figure(figsize=(10, 10))
plt.imshow(img_act_rgb)
plt.title("7. Rectangle and Circle on Face")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("/content/drive/MyDrive/PCVK_2026/Student_KTM.jpeg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

h, w, _ = img_rgb.shape

# Create blue channel image
img_blue = img_rgb.copy()

# Box size
box_size = 100

# Draw rectangle in the lower-left corner
cv2.rectangle(
    img_blue,
    (10, h - box_size),
    (10 + box_size, h),
    (0, 0, 255),
    3
)

# Add filename
filename = "Student_KTM.jpeg"

cv2.putText(
    img_blue,
    filename,
    (10, h - box_size - 15),
    cv2.FONT_HERSHEY_DUPLEX,
    0.8,
    (0, 255, 255),
    2,
    cv2.LINE_AA
)

# Display the image
plt.figure(figsize=(10, 6))
plt.imshow(img_blue)
plt.title("9. Filename on Image")
plt.axis("off")
plt.show()
```

```python
img = cv2.imread("/content/drive/MyDrive/PCVK_2026/Student_KTM.jpeg")

# Coordinates for the face in your KTM photo
left_eye_center = (350, 535)
right_eye_center = (385, 535)

mouth_top_left = (350, 585)
mouth_bottom_right = (395, 605)

# Draw circles on the eyes
cv2.circle(img, left_eye_center, radius=14, color=(0, 0, 255), thickness=3)
cv2.circle(img, right_eye_center, radius=14, color=(0, 255, 0), thickness=3)

# Draw rectangle around the mouth
cv2.rectangle(
    img,
    mouth_top_left,
    mouth_bottom_right,
    color=(255, 0, 0),
    thickness=3
)

# Convert BGR to RGB
img_RGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 6))
plt.imshow(img_RGB)
plt.axis("on")
plt.show()
```
